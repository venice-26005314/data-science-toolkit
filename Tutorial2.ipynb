```python
#Part 1-Google Colab and Python
# 1.Create a new notebook and code the following
# 1.Create a numpy array of size 2x2

import numpy as np

array_2x2=np.array([
    [1,2],
    [3,4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4

random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes

ones_array=np.ones((3,3))
zeros_array=np.zeros((3,3))

print("Array of Ones:")
print(ones_array)

print("Array of zeros:")
print(zeros_array)
```

```python
#4. Create a random numpy array of size 6x3 and transpose it

array_6x3=np.random.rand(6,3)
transposed_array=np.transpose(array_6x3)

print("Original 6x3 array:")
print(array_6x3)

print("Transposed array:")
print(transposed_array)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product

array_4x2=np.random.rand(4,2)
array_2x3=np.random.rand(2,3)

dot_product=np.dot(array_4x2,array_2x3)

print("Dot product of 4x2 and 2x3:")
print(dot_product)
```
